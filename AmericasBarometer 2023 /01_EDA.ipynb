### Nota sobre recodificaciones pendientes

Antes del análisis, varias variables requieren transformación:

- **`pn4`** tiene escala invertida respecto a la intuición: 1 es positivo y 4 es negativo. Para la regresión logística se recodifica como binaria: `pn4_bin = 1` si pn4 ∈ {1,2}, `pn4_bin = 0` si pn4 ∈ {3,4}.
- **`pol1`**, **`m1`** y **`aoj11`** también tienen escala donde valores bajos son positivos. Conviene invertirlas para que valores altos siempre signifiquen "más" de lo que mide la variable, facilitando la interpretación de los coeficientes.
- **`b1` a `b6`** y **`d1` a `d4`** se van a promediar para crear índices. Antes hay que asegurarse de que estén en la misma escala y excluir valores faltantes.

## Decisiones sobre valores faltantes

El análisis de valores faltantes muestra una situación favorable: ninguna variable
supera el 10% de datos faltantes.

**Decisión adoptada: listwise deletion**  
Para todos los modelos se excluirán los casos con valores faltantes en las variables
incluidas. Dado que los porcentajes son bajos (máximo 9.8% en `ingreso`), esta
estrategia no introduce sesgos significativos.

**Variable con mayor ausencia:**
- `ingreso` (9.8%): patrón típico en encuestas de opinión pública.
  Se conserva en el análisis con la advertencia de que los modelos
  que la incluyan operarán con una n ligeramente menor.

**Variables dependientes principales:**
- `pn4` (satisfacción con la democracia): 2.2% faltante → n efectiva ≈ 1494
- `ing4` (preferencia por la democracia): 1.7% faltante → n efectiva ≈ 1501

```python
# Distribución de pn4 - Satisfacción con la democracia
# Excluir valores faltantes
pn4_limpia = df_trabajo['pn4'].dropna()

# Conteo y porcentajes
conteo = pn4_limpia.value_counts().sort_index()
porcentaje = (pn4_limpia.value_counts(normalize=True) * 100).sort_index().round(1)

# Tabla resumen
resumen_pn4 = pd.DataFrame({
    'Respuesta': ['Muy satisfecho', 'Satisfecho', 'Insatisfecho', 'Muy insatisfecho'],
    'N': conteo.values,
    'Porcentaje': porcentaje.values
})
print(resumen_pn4)
print(f"\n→ Satisfechos (1+2): {porcentaje[1.0] + porcentaje[2.0]:.1f}%")
print(f"→ Informe oficial:   61.0%")
```

```python
# Gráfico
etiquetas = ['Muy\nsatisfecho', 'Satisfecho', 'Insatisfecho', 'Muy\ninsatisfecho']
colores = ['#2ecc71', '#82e0aa', '#e59866', '#e74c3c']

fig, ax = plt.subplots(figsize=(9, 5))

barras = ax.bar(etiquetas, porcentaje.values, color=colores, width=0.5)

# Etiquetas encima de cada barra
for barra, pct in zip(barras, porcentaje.values):
    ax.text(barra.get_x() + barra.get_width() / 2,
            barra.get_height() + 0.5,
            f'{pct}%',
            ha='center', va='bottom', fontsize=11)

ax.set_title('Satisfacción con el funcionamiento de la democracia\nCosta Rica 2023',
             fontsize=13, pad=15)
ax.set_ylabel('Porcentaje (%)')
ax.set_ylim(0, max(porcentaje.values) + 8)

# Línea que separa satisfechos de insatisfechos
ax.axvline(x=1.5, color='gray', linestyle='--', alpha=0.5)
ax.text(0.75, max(porcentaje.values) + 5, 'Satisfechos', 
        ha='center', color='#27ae60', fontsize=10)
ax.text(2.25, max(porcentaje.values) + 5, 'Insatisfechos', 
        ha='center', color='#c0392b', fontsize=10)

plt.tight_layout()
plt.savefig('../outputs/01_satisfaccion_democracia.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Validación: Satisfacción con la democracia

**Resultado obtenido:** 61.4% satisfechos (Muy satisfecho + Satisfecho)  
**Informe oficial LAPOP 2023:** 61%  
**Diferencia:** 0.4 puntos porcentuales ✓

La replicación es exitosa. La pequeña diferencia se debe a que el informe
oficial usó una versión preliminar del dataset con actividades de limpieza
y control de calidad adicionales, como indica la ficha técnica del informe.

**Lectura sustantiva:** La satisfacción es predominantemente moderada.
El 52.1% está "satisfecho" pero solo el 9.3% está "muy satisfecho",
lo que sugiere un respaldo frágil, condicionado al desempeño institucional.
Esto es consistente con lo que Alfaro Redondo describe como una democracia
resiliente pero amenazada.

```python
# Distribución de ing4 - Preferencia por la democracia
# ing4 va de 1 (muy en desacuerdo) a 7 (muy de acuerdo)
ing4_limpia = df_trabajo['ing4'].dropna()

# Porcentaje por valor
porcentaje_ing4 = (ing4_limpia.value_counts(normalize=True) * 100).sort_index().round(1)

print("Distribución de ing4:")
print(porcentaje_ing4)

# El informe recodifica: valores 5,6,7 = "de acuerdo" con la democracia
pct_favor = ing4_limpia[ing4_limpia >= 5].count() / len(ing4_limpia) * 100
print(f"\n→ De acuerdo con la democracia (5-7): {pct_favor:.1f}%")
print(f"→ Informe oficial:                    72.0%")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Gráfico izquierdo: distribución completa ---
colores_ing4 = ['#e74c3c', '#e67e22', '#f39c12', 
                '#95a5a6', '#82e0aa', '#2ecc71', '#1a9850']

axes[0].bar(porcentaje_ing4.index.astype(int), 
            porcentaje_ing4.values,
            color=colores_ing4, width=0.6)

# Etiquetas
for x, y in zip(porcentaje_ing4.index.astype(int), porcentaje_ing4.values):
    axes[0].text(x, y + 0.3, f'{y}%', ha='center', fontsize=9)

axes[0].set_xticks(range(1, 8))
axes[0].set_xticklabels(['1\nMuy en\ndesacuerdo', '2', '3', '4', '5', '6', 
                          '7\nMuy de\nacuerdo'], fontsize=8)
axes[0].set_ylabel('Porcentaje (%)')
axes[0].set_title('Distribución completa\n"La democracia es mejor que\ncualquier otra forma de gobierno"')
axes[0].axvline(x=4.5, color='gray', linestyle='--', alpha=0.5)

# --- Gráfico derecho: comparación con informe oficial ---
categorias = ['De acuerdo\n(valores 5-7)', 'En desacuerdo\n(valores 1-4)']
valores = [pct_favor, 100 - pct_favor]
colores_comp = ['#2ecc71', '#e74c3c']

barras = axes[1].bar(categorias, valores, color=colores_comp, width=0.4)

for barra, val in zip(barras, valores):
    axes[1].text(barra.get_x() + barra.get_width() / 2,
                 barra.get_height() + 0.5,
                 f'{val:.1f}%', ha='center', fontsize=12)

# Línea del informe oficial
axes[1].axhline(y=72, color='navy', linestyle='--', 
                linewidth=1.5, label='Informe oficial (72%)')
axes[1].legend()
axes[1].set_ylabel('Porcentaje (%)')
axes[1].set_ylim(0, 100)
axes[1].set_title('Comparación con\ninforme oficial LAPOP 2023')

plt.suptitle('Preferencia por la democracia — Costa Rica 2023', 
             fontsize=13, y=1.02)
plt.tight_layout()
plt.savefig('../outputs/02_preferencia_democracia.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Validación: Preferencia por la democracia

**Resultado obtenido:** 72.1% de acuerdo con que la democracia es mejor
que cualquier otra forma de gobierno (valores 5-7 en escala 1-7)  
**Informe oficial LAPOP 2023:** 72%

```python
# Todas las variables de confianza institucional juntas
confianza_vars = ['b1', 'b2', 'b3', 'b4', 'b6',
                  'b13', 'b18', 'b21', 'b31', 'b21a', 'b47a']

etiquetas_confianza = {
    'b1':   'Tribunales',
    'b2':   'Inst. políticas',
    'b3':   'Derechos',
    'b4':   'Orgullo sistema',
    'b6':   'Apoyar sistema',
    'b13':  'Asamblea',
    'b18':  'Policía',
    'b21':  'Partidos',
    'b31':  'Corte Suprema',
    'b21a': 'Presidente',
    'b47a': 'Elecciones'
}

# Calcular correlaciones
df_confianza = df_trabajo[confianza_vars].rename(columns=etiquetas_confianza)
correlaciones = df_confianza.corr()

# Heatmap
fig, ax = plt.subplots(figsize=(11, 9))

mascara = np.zeros_like(correlaciones, dtype=bool)
mascara[np.triu_indices_from(mascara)] = True

sns.heatmap(correlaciones,
            mask=mascara,
            annot=True,
            fmt='.2f',
            cmap='RdYlGn',
            vmin=0, vmax=1,
            square=True,
            linewidths=0.5,
            ax=ax,
            annot_kws={'size': 8})

ax.set_title('Correlaciones entre variables de confianza institucional\nCosta Rica 2023',
             fontsize=13, pad=15)

plt.tight_layout()
plt.savefig('../outputs/04_heatmap_confianza.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Satisfacción con la democracia según perfil sociodemográfico

Exploramos cómo varía la satisfacción con la democracia (pn4) según
género, edad, educación, zona de residencia e ingreso familiar.

Para este análisis recodificamos pn4 como variable binaria:
- 1 = Satisfecho (muy satisfecho + satisfecho)
- 0 = Insatisfecho (insatisfecho + muy insatisfecho)

Esto permite calcular el porcentaje de satisfechos en cada subgrupo
y comparar directamente con el 61% general reportado por el informe oficial.

```python
# Recodificación de pn4
# 1 y 2 = satisfecho → 1
# 3 y 4 = insatisfecho → 0
df_trabajo['pn4_bin'] = df_trabajo['pn4'].map({
    1.0: 1, 2.0: 1,
    3.0: 0, 4.0: 0
})

# Verificar
print("Distribución de pn4_bin:")
print(df_trabajo['pn4_bin'].value_counts())
print(f"\nPorcentaje satisfechos: {df_trabajo['pn4_bin'].mean()*100:.1f}%")
```

```python
# Función reutilizable para calcular % satisfechos por subgrupo
def pct_satisfechos(df, variable):
    resultado = (df.groupby(variable)['pn4_bin']
                   .agg(['mean', 'count'])
                   .reset_index())
    resultado['pct'] = (resultado['mean'] * 100).round(1)
    return resultado
```

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

# Línea de referencia: promedio general
promedio_general = df_trabajo['pn4_bin'].mean() * 100

# ── Gráfico 1: Género ──
genero_df = df_trabajo[df_trabajo['genero'].isin([1.0, 2.0])]
res_genero = pct_satisfechos(genero_df, 'genero')
etiq_genero = {1.0: 'Hombre', 2.0: 'Mujer'}

axes[0].bar(
    [etiq_genero[v] for v in res_genero['genero']],
    res_genero['pct'],
    color=['#3498db', '#e91e8c'],
    width=0.4
)
for i, row in res_genero.iterrows():
    axes[0].text(i, row['pct'] + 0.5, f"{row['pct']}%",
                 ha='center', fontsize=10)
axes[0].axhline(promedio_general, color='gray',
                linestyle='--', linewidth=1.2,
                label=f'Promedio: {promedio_general:.1f}%')
axes[0].set_title('Por género', fontsize=11)
axes[0].set_ylabel('% satisfechos')
axes[0].set_ylim(0, 85)
axes[0].legend(fontsize=8)

# ── Gráfico 2: Zona ──
res_zona = pct_satisfechos(df_trabajo, 'zona')
etiq_zona = {1: 'Urbano', 2: 'Rural'}

axes[1].bar(
    [etiq_zona[v] for v in res_zona['zona']],
    res_zona['pct'],
    color=['#2ecc71', '#f39c12'],
    width=0.4
)
for i, row in res_zona.iterrows():
    axes[1].text(i, row['pct'] + 0.5, f"{row['pct']}%",
                 ha='center', fontsize=10)
axes[1].axhline(promedio_general, color='gray',
                linestyle='--', linewidth=1.2)
axes[1].set_title('Por zona de residencia', fontsize=11)
axes[1].set_ylim(0, 85)

# ── Gráfico 3: Educación ──
res_edu = pct_satisfechos(df_trabajo, 'educacion')
etiq_edu = {
    0.0: 'Ninguna',
    1.0: 'Prim.\nIncomp.',
    2.0: 'Prim.\nComp.',
    3.0: 'Sec.\nIncomp.',
    4.0: 'Sec.\nComp.',
    5.0: 'Sup.\nIncomp.',
    6.0: 'Sup.\nComp.'
}

axes[2].bar(
    [etiq_edu[v] for v in res_edu['educacion']],
    res_edu['pct'],
    color='#9b59b6',
    alpha=0.85
)
for i, row in res_edu.iterrows():
    axes[2].text(i, row['pct'] + 0.5, f"{row['pct']}%",
                 ha='center', fontsize=8)
axes[2].axhline(promedio_general, color='gray',
                linestyle='--', linewidth=1.2)
axes[2].set_title('Por nivel educativo', fontsize=11)
axes[2].set_ylabel('% satisfechos')
axes[2].set_ylim(0, 85)
axes[2].tick_params(axis='x', labelsize=8)

# ── Gráfico 4: Grupos de edad ──
# Crear grupos de edad
bins_edad = [17, 25, 35, 45, 55, 65, 100]
etiq_edad = ['18-25', '26-35', '36-45', '46-55', '56-65', '66+']
df_trabajo['grupo_edad'] = pd.cut(df_trabajo['edad'],
                                   bins=bins_edad,
                                   labels=etiq_edad)

res_edad = pct_satisfechos(df_trabajo, 'grupo_edad')

axes[3].bar(
    res_edad['grupo_edad'].astype(str),
    res_edad['pct'],
    color='steelblue',
    alpha=0.85
)
for i, row in res_edad.iterrows():
    axes[3].text(i, row['pct'] + 0.5, f"{row['pct']}%",
                 ha='center', fontsize=9)
axes[3].axhline(promedio_general, color='gray',
                linestyle='--', linewidth=1.2)
axes[3].set_title('Por grupo de edad', fontsize=11)
axes[3].set_ylabel('% satisfechos')
axes[3].set_ylim(0, 85)

# ── Gráfico 5: Ingreso ──
etiq_ingreso = {
    0.0:  'Sin\ningreso',
    1.0:  'Menos\n₡60k',
    2.0:  '₡60k-\n105k',
    3.0:  '₡105k-\n136k',
    4.0:  '₡136k-\n170k',
    5.0:  '₡170k-\n210k',
    6.0:  '₡210k-\n240k',
    7.0:  '₡240k-\n275k',
    8.0:  '₡275k-\n310k',
    9.0:  '₡310k-\n345k',
    10.0: '₡345k-\n415k',
    11.0: '₡415k-\n490k',
    12.0: '₡490k-\n560k',
    13.0: '₡560k-\n715k',
    14.0: '₡715k-\n900k',
    15.0: '₡900k-\n1.2M',
    16.0: 'Más de\n₡1.2M'
}

# Agrupar en quintiles para que el gráfico sea legible
df_trabajo['quintil_ingreso'] = pd.qcut(
    df_trabajo['ingreso'].dropna(),
    q=5,
    labels=['Q1\n(más bajo)', 'Q2', 'Q3', 'Q4', 'Q5\n(más alto)']
)

res_ingreso = (df_trabajo.groupby('quintil_ingreso', observed=True)['pn4_bin']
               .agg(['mean', 'count'])
               .reset_index())
res_ingreso['pct'] = (res_ingreso['mean'] * 100).round(1)

colores_ingreso = ['#e74c3c', '#e67e22', '#f1c40f', '#2ecc71', '#27ae60']

axes[4].bar(
    res_ingreso['quintil_ingreso'].astype(str),
    res_ingreso['pct'],
    color=colores_ingreso,
    alpha=0.9
)
for i, row in res_ingreso.iterrows():
    axes[4].text(i, row['pct'] + 0.5, f"{row['pct']}%",
                 ha='center', fontsize=9)
axes[4].axhline(promedio_general, color='gray',
                linestyle='--', linewidth=1.2)
axes[4].set_title('Por quintil de ingreso', fontsize=11)
axes[4].set_ylim(0, 85)

# ── Eliminar el sexto subplot vacío ──
fig.delaxes(axes[5])

plt.suptitle('Satisfacción con la democracia según perfil sociodemográfico\nCosta Rica 2023',
             fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig('../outputs/06_satisfaccion_por_grupo.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Variables y etiquetas
vars_mig = {
    'inm_quitan_empleo_r': 'No quitan\nempleos',
    'nic_empeoran_crimen_r': 'No empeoran\nel crimen',
    'nic_economia_r': 'Buenos para\nla economía',
    'nic_cultura_r': 'Enriquecen\nla cultura',
    'inm_derechos_r': 'Mismos\nderechos'
}

# Tres variables sociodemográficas
sociodem = {
    'grupo_edad': {
        'titulo': 'Por grupo de edad',
        'color': 'steelblue'
    },
    'educacion': {
        'titulo': 'Por nivel educativo',
        'color': '#9b59b6'
    },
    'zona': {
        'titulo': 'Por zona de residencia',
        'color': '#2ecc71'
    }
}

etiq_zona = {1: 'Urbano', 2: 'Rural'}
etiq_edu = {
    0.0: 'Ninguna',
    1.0: 'Prim.\nInc.',
    2.0: 'Prim.\nComp.',
    3.0: 'Sec.\nInc.',
    4.0: 'Sec.\nComp.',
    5.0: 'Sup.\nInc.',
    6.0: 'Sup.\nComp.'
}

fig, axes = plt.subplots(5, 3, figsize=(18, 22))

for fila, (var_mig, label_mig) in enumerate(vars_mig.items()):
    for col, (var_soc, config) in enumerate(sociodem.items()):

        ax = axes[fila, col]
        df_temp = df_trabajo[[var_soc, var_mig]].dropna()

        # Calcular porcentaje positivo por grupo
        resultado = (df_temp.groupby(var_soc)[var_mig]
                     .agg(['mean', 'count'])
                     .reset_index())
        resultado['pct'] = (resultado['mean'] * 100).round(1)

        # Etiquetas del eje x según variable
        if var_soc == 'zona':
            xtick_labels = [etiq_zona.get(v, str(v)) 
                           for v in resultado[var_soc]]
        elif var_soc == 'educacion':
            xtick_labels = [etiq_edu.get(v, str(v)) 
                           for v in resultado[var_soc]]
        else:
            xtick_labels = resultado[var_soc].astype(str).tolist()

        # Graficar
        barras = ax.bar(range(len(resultado)),
                        resultado['pct'],
                        color=config['color'],
                        alpha=0.8,
                        edgecolor='white')

        # Etiquetas encima de barras
        for i, row in resultado.iterrows():
            ax.text(i, row['pct'] + 0.8,
                    f"{row['pct']}%",
                    ha='center', fontsize=7.5)

        # Línea de promedio general
        promedio = df_trabajo[var_mig].mean() * 100
        ax.axhline(promedio, color='red',
                   linestyle='--', linewidth=1,
                   label=f'Promedio: {promedio:.1f}%')

        ax.set_xticks(range(len(resultado)))
        ax.set_xticklabels(xtick_labels, fontsize=7)
        ax.set_ylim(0, 100)
        ax.set_ylabel('% actitud positiva' if col == 0 else '')
        ax.legend(fontsize=7)

        # Títulos
        if fila == 0:
            ax.set_title(config['titulo'], fontsize=11, pad=10)
        if col == 0:
            ax.set_ylabel(f"{label_mig}\n% positivo", fontsize=9)

plt.suptitle('Actitudes hacia la migración según perfil sociodemográfico\nCosta Rica 2023',
             fontsize=14, y=1.01)
plt.tight_layout()
plt.savefig('../outputs/07_migracion_por_grupo.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 7. Decisiones metodológicas para los módulos siguientes

Con base en el EDA, se adoptan las siguientes decisiones:
